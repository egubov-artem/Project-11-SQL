# Проект "DonorSearch"

## Описание проекта

В этой работе будут использованы данные проекта DonorSearch. Цель проекта — мотивировать людей становиться донорами и делать регулярные донации. Для этого важно понимать, какие факторы влияют на активность доноров и какими могут быть стратегии для их мотивации.

## Цель и задачи проекта

Цель проекта - выработать подходы, которые помогут проекту DonorSearch достигать своих целей, а именно мотивировать людей становиться донорами и делать регулярные донации.

Задачи, поставленные в проекте:
- Определить регионы с наибольшим количеством зарегистрированных доноров.
- Изучить динамику общего количества донаций в месяц за 2022 и 2023 годы.
- Определить наиболее активных доноров в системе, учитывая только данные о зарегистрированных и подтвержденных донациях.
- Оценить, как система бонусов влияет на зарегистрированные в системе донации.
- Исследовать вовлечение новых доноров через социальные сети. Узнать, сколько по каким каналам пришло доноров, и среднее количество донаций по каждому каналу.
- Сравнить активность однократных доноров со средней активностью повторных доноров.
- Сравнить данные о планируемых донациях с фактическими данными, чтобы оценить эффективность планирования.

Найдено большое количество дубликатов, следует проверить, чем они отличаются кроме id на примере одного пользователя:

```python
%%sql

SELECT *
FROM donorsearch.donation_anon
WHERE user_id = 211970
  AND donation_date = '2010-03-13'
  AND donation_type = 'Безвозмездно'
LIMIT 10;
```

Записи этого пользователя по связке 3 полей полностью идентичны (не считая id). Проверим другие такие связки: отличаются ли они по другим полям.

```python
%%sql

WITH dups AS (
  SELECT user_id, donation_date, donation_type
  FROM donorsearch.donation_anon
  GROUP BY 1, 2, 3
  HAVING COUNT(*) > 1
)
SELECT
  COUNT(*)                                    AS dup_groups,
  COUNT(*) FILTER (WHERE n_blood > 1)         AS differ_blood_class,
  COUNT(*) FILTER (WHERE n_place > 1)         AS differ_place,
  COUNT(*) FILTER (WHERE n_conf > 1)          AS differ_confirmation,
  COUNT(*) FILTER (WHERE n_status > 1)        AS differ_status,
  COUNT(*) FILTER (WHERE n_added > 1)         AS differ_added_date
FROM (
  SELECT a.user_id, a.donation_date, a.donation_type,
         COUNT(DISTINCT blood_class)          AS n_blood,
         COUNT(DISTINCT donation_place)       AS n_place,
         COUNT(DISTINCT confirmation)         AS n_conf,
         COUNT(DISTINCT donation_status)      AS n_status,
         COUNT(DISTINCT donation_added_date)  AS n_added
  FROM donorsearch.donation_anon a
  JOIN dups d USING (user_id, donation_date, donation_type)
  GROUP BY a.user_id, a.donation_date, a.donation_type
) t;
```

Группы дубликатов с полностью идентичным содержимым (кроме id) составляют меньшинство.

Проверим дальше, какие бывают статусы и сколько строк в каждой связке donation_status - confirmation.

```python
%%sql

SELECT donation_status, confirmation, COUNT(*) AS cnt
FROM donorsearch.donation_anon
GROUP BY 1, 2
ORDER BY cnt DESC;
```

Выводы:

Теперь можно посмотреть сколько останется после очистки и как изменится картина дублей.

```python
%%sql

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           CASE donation_status WHEN 'Принята' THEN 1 ELSE 2 END,
           confirmation DESC NULLS LAST, id
)
SELECT
  (SELECT COUNT(*) FROM donorsearch.donation_anon) AS raw_rows,
  COUNT(*)                                        AS clean_rows,
  COUNT(*) FILTER (WHERE donation_status = 'Принята') AS accepted_only,
  COUNT(DISTINCT user_id)                         AS donors
FROM clean;
```

Выводы:
- Очистка убрала 43 424 строк (17,7%). Это сумма нескольких частей: донации без справки, удаленные, отклоненные и на модерациие и то, что приходится на дедупликацию.
- В очищенной версии остануться строки со статусом "Принята" и обоими значениями confirmation.

Проверим, остались ли дубликаты после очистки.

```python
%%sql

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           CASE donation_status WHEN 'Принята' THEN 1 ELSE 2 END,
           confirmation DESC NULLS LAST, id
)
SELECT COUNT(*) - COUNT(DISTINCT (user_id, donation_date, donation_type, blood_class)) AS dups_left
FROM clean;
```

Теперь проверим согласуется ли очистка с полем платформы `confirmed_donations`

```python
%%sql

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
cnt AS (SELECT user_id, COUNT(*) AS n FROM clean GROUP BY user_id)
SELECT
  COUNT(*) AS users,
  COUNT(*) FILTER (WHERE u.confirmed_donations = COALESCE(c.n, 0)) AS match_clean,
  COUNT(*) FILTER (WHERE u.confirmed_donations <> COALESCE(c.n, 0)) AS mismatch
FROM donorsearch.user_anon_data u
LEFT JOIN cnt c ON c.user_id = u.id;
```

Логика подсчета зарегистрированных и подтвержденных заявок совпадает на 99,4%.

Выводы:
- Успешная донация определена как запись со статусом "Принята", один раз на комбинацию (`user_id`, `donation_date`, `donation_type`, `blood_class`)
- Из 245 744 сырых строк в основную метрику вошло 202 320 (82,3%) - это зарегистрированные, подтвержденные донации
- Присутствует расхождение с платформенным `confirmed_donations` у 3,6% доноров с донациями, что можно считать несущественным

## 2. Регионы с наибольшим количеством зарегистрированных доноров

```python
%%sql
-- Сначала посмотрим, нужна ли нормализация.

SELECT
  COUNT(DISTINCT region)                    AS distinct_raw,
  COUNT(DISTINCT LOWER(TRIM(region)))       AS distinct_normalized,
  COUNT(*) FILTER (WHERE region IS NULL)    AS null_region,
  COUNT(*) FILTER (WHERE TRIM(region) = '') AS empty_region
FROM donorsearch.user_anon_data;
```

```python
%%sql
-- Варианты, различающиеся только регистром или пробелами.

SELECT LOWER(TRIM(region)) AS norm,
       ARRAY_AGG(DISTINCT region) AS variants,
       COUNT(*) AS donors
FROM donorsearch.user_anon_data
WHERE region IS NOT NULL
GROUP BY 1
HAVING COUNT(DISTINCT region) > 1
ORDER BY donors DESC;
```

- Две пары различаются только пробелом на конце. Нормализация TRIM это исправляет.

```python
%%sql
-- Просмотр самых частых значений.

SELECT region, COUNT(*) AS donors
FROM donorsearch.user_anon_data
GROUP BY region
ORDER BY donors DESC
LIMIT 5;
```

- Самый частый вариант - отсутствие записи о регионе, его следует исключить из топа.
- Стоит заметить, что поле `region` не является регионом, Значения имеют формат "Страна, Регион, Город".
- Если считать GROUP BY `region` как есть, получится рейтинг городов.
- Составим отдельно топ городов и топ регионов.

```python
%%sql
-- Распределение доноров по форме записи региона

SELECT LENGTH(region) - LENGTH(REPLACE(region, ',', '')) AS n_commas,
       COUNT(*) AS donors
FROM donorsearch.user_anon_data
WHERE region IS NOT NULL
GROUP BY 1
ORDER BY 1;
```

```python
%%sql
-- Топ-15 регионов по числу доноров

WITH parsed AS (
  SELECT
    id,
    NULLIF(NULLIF(TRIM(SPLIT_PART(region, ',', 1)), ''), '_') AS country,
    LENGTH(region) - LENGTH(REPLACE(region, ',', '')) AS n_commas,
    TRIM(SPLIT_PART(region, ',', 2)) AS part2
  FROM donorsearch.user_anon_data
  WHERE region IS NOT NULL
),
reg AS (
  SELECT id, country,
    CASE
      WHEN n_commas >= 2 THEN NULLIF(NULLIF(part2, ''), '_')
      WHEN n_commas = 1 AND part2 IN ('Москва', 'Санкт-Петербург', 'Севастополь') THEN part2
      ELSE NULL
    END AS region_name
  FROM parsed
),
totals AS (
  SELECT COUNT(*)                                          AS known_donors,
         (SELECT COUNT(*) FROM donorsearch.user_anon_data) AS all_donors
  FROM parsed
)
SELECT
  r.country,
  r.region_name,
  COUNT(*)                                      AS donors,
  ROUND(100.0 * COUNT(*) / t.known_donors, 2)   AS pct_of_known,
  ROUND(100.0 * COUNT(*) / t.all_donors, 2)     AS pct_of_all
FROM reg r
CROSS JOIN totals t
WHERE r.region_name IS NOT NULL
GROUP BY r.country, r.region_name, t.known_donors, t.all_donors
ORDER BY donors DESC
LIMIT 15;
```

```python
%%sql
-- Топ-15 городов по числу доноров

WITH parsed AS (
  SELECT
    id,
    NULLIF(NULLIF(TRIM(SPLIT_PART(region, ',', 1)), ''), '_') AS country,
    LENGTH(region) - LENGTH(REPLACE(region, ',', '')) AS n_commas,
    CASE
      WHEN LENGTH(region) - LENGTH(REPLACE(region, ',', '')) = 1
        THEN TRIM(SPLIT_PART(region, ',', 2))
      WHEN LENGTH(region) - LENGTH(REPLACE(region, ',', '')) = 2
        THEN TRIM(SPLIT_PART(region, ',', 3))
      ELSE TRIM(SPLIT_PART(region, ',', 3) || ',' || SPLIT_PART(region, ',', 4))
    END AS city
  FROM donorsearch.user_anon_data
  WHERE region IS NOT NULL
),
totals AS (
  SELECT COUNT(*) AS known_donors,
         (SELECT COUNT(*) FROM donorsearch.user_anon_data) AS all_donors
  FROM parsed
)
SELECT
  p.country,
  p.city,
  COUNT(*)                                     AS donors,
  ROUND(100.0 * COUNT(*) / t.known_donors, 2)  AS pct_of_known,
  ROUND(100.0 * COUNT(*) / t.all_donors, 2)    AS pct_of_all
FROM parsed p
CROSS JOIN totals t
WHERE p.city <> ''
GROUP BY p.country, p.city, t.known_donors, t.all_donors
ORDER BY donors DESC
LIMIT 15;
```

```python
%%sql
-- Доля неизвестного региона

SELECT
  COUNT(*) FILTER (WHERE NULLIF(TRIM(region), '') IS NULL) AS unknown_region,
  COUNT(*)                                                 AS total,
  ROUND(100.0 * COUNT(*) FILTER (WHERE NULLIF(TRIM(region), '') IS NULL) / COUNT(*), 1) AS pct_unknown
FROM donorsearch.user_anon_data;
```

## 3. Динамика общего количества донаций в месяц за 2022 и 2023 годы

```python
%%sql
-- Донации по месяцам

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
)
SELECT
  DATE_TRUNC('month', donation_date)::date AS month,
  COUNT(*)                                 AS donations,
  COUNT(DISTINCT user_id)                  AS donors
FROM clean
WHERE donation_date >= '2022-01-01'
  AND donation_date <= '2023-11-28'
GROUP BY 1
ORDER BY 1;
```

```python
%%sql
-- Сравнение аналогичных месяцев

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
m AS (
  SELECT EXTRACT(YEAR FROM donation_date)::int  AS yr,
         EXTRACT(MONTH FROM donation_date)::int AS mon,
         COUNT(*) AS donations
  FROM clean
  WHERE donation_date >= '2022-01-01'
  GROUP BY 1, 2
)
SELECT
  mon,
  MAX(donations) FILTER (WHERE yr = 2022) AS d_2022,
  MAX(donations) FILTER (WHERE yr = 2023) AS d_2023,
  MAX(donations) FILTER (WHERE yr = 2023) - MAX(donations) FILTER (WHERE yr = 2022) AS diff,
  ROUND(100.0 * (MAX(donations) FILTER (WHERE yr = 2023)
               - MAX(donations) FILTER (WHERE yr = 2022))
        / MAX(donations) FILTER (WHERE yr = 2022), 1) AS pct_change
FROM m
GROUP BY mon
ORDER BY mon;
```

```python
%%sql
-- Проверка: не вызван ли рост массовой загрузкой

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
)
SELECT
  DATE_TRUNC('month', donation_added_date)::date AS added_month,
  COUNT(*) AS rows_added,
  COUNT(*) FILTER (WHERE donation_date >= '2022-01-01') AS of_which_2022_2023
FROM clean
GROUP BY 1
ORDER BY rows_added DESC
LIMIT 12;
```

```python
%%sql
-- Разрезы по типу донации и по компоненту крови
    
WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
)
SELECT
  EXTRACT(YEAR FROM donation_date)::int AS yr,
  donation_type,
  blood_class,
  COUNT(*) AS donations
FROM clean
WHERE donation_date >= '2022-01-01'
GROUP BY 1, 2, 3
ORDER BY yr, donations DESC;
```

```python
%%sql
-- Зависимость от числа взятых бонусов

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
per_user AS (SELECT user_id, COUNT(*) AS n FROM clean GROUP BY 1),
b AS (SELECT user_id, COUNT(*) AS bonuses FROM donorsearch.user_anon_bonus GROUP BY 1)
SELECT
  CASE WHEN b.bonuses IS NULL THEN '0'
       WHEN b.bonuses = 1 THEN '1'
       WHEN b.bonuses BETWEEN 2 AND 3 THEN '2-3'
       WHEN b.bonuses BETWEEN 4 AND 10 THEN '4-10'
       ELSE '10+' END                              AS bonuses_bucket,
  COUNT(*)                                         AS donors,
  ROUND(AVG(p.n), 1)                               AS avg_donations,
  PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY p.n) AS median_donations
FROM per_user p
LEFT JOIN b ON b.user_id = p.user_id
GROUP BY 1
ORDER BY MIN(COALESCE(b.bonuses, 0));
```

```python
%%sql
-- До и после первого бонуса

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
first_bonus AS (
  SELECT user_id, MIN(date_of_use) AS fb
  FROM donorsearch.user_anon_bonus
  WHERE date_of_use IS NOT NULL
  GROUP BY 1
  HAVING MIN(date_of_use) BETWEEN '2021-06-01' AND '2023-05-31'
),
w AS (
  SELECT f.user_id,
         COUNT(*) FILTER (WHERE d.donation_date >= f.fb - 180 AND d.donation_date <  f.fb)       AS before_n,
         COUNT(*) FILTER (WHERE d.donation_date >= f.fb       AND d.donation_date <  f.fb + 180) AS after_n
  FROM first_bonus f
  LEFT JOIN clean d ON d.user_id = f.user_id
  GROUP BY f.user_id
)
SELECT
  COUNT(*)                                            AS donors,
  ROUND(AVG(before_n), 2)                             AS avg_before,
  ROUND(AVG(after_n), 2)                              AS avg_after,
  COUNT(*) FILTER (WHERE after_n > before_n)          AS more_after,
  COUNT(*) FILTER (WHERE after_n < before_n)          AS fewer_after,
  COUNT(*) FILTER (WHERE after_n = before_n)          AS same,
  ROUND(100.0 * COUNT(*) FILTER (WHERE after_n = 0) / COUNT(*), 1) AS pct_no_donation_after
FROM w;
```

```python
%%sql
-- Контрольная группа: когорта на 01.06.2022

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
idx AS (SELECT DATE '2022-06-01' AS d),
active AS (
  SELECT c.user_id, COUNT(*) AS n_before
  FROM clean c, idx
  WHERE c.donation_date >= idx.d - 365 AND c.donation_date < idx.d
  GROUP BY c.user_id
),
grp AS (
  SELECT a.user_id, a.n_before,
         EXISTS (SELECT 1 FROM donorsearch.user_anon_bonus b, idx
                 WHERE b.user_id = a.user_id AND b.date_of_use < idx.d) AS took_bonus
  FROM active a
),
outcome AS (
  SELECT g.user_id, g.took_bonus, g.n_before,
         COUNT(c.id) AS n_after
  FROM grp g
  CROSS JOIN idx
  LEFT JOIN clean c
         ON c.user_id = g.user_id
        AND c.donation_date >= idx.d AND c.donation_date < idx.d + 180
  GROUP BY g.user_id, g.took_bonus, g.n_before
)
SELECT
  CASE WHEN n_before = 1 THEN '1'
       WHEN n_before <= 3 THEN '2-3'
       ELSE '4+' END                                      AS prior_year_bucket,
  took_bonus,
  COUNT(*)                                                AS donors,
  ROUND(AVG(n_after), 2)                                  AS avg_donations_after,
  ROUND(100.0 * COUNT(*) FILTER (WHERE n_after > 0) / COUNT(*), 1) AS pct_returned
FROM outcome
GROUP BY 1, 2
ORDER BY 1, 2;
```

```python
%%sql
-- Какие бонусы связаны с активностью

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
per_user AS (SELECT user_id, COUNT(*) AS n FROM clean GROUP BY 1)
SELECT
  b.partner,
  b.bonus_name,
  COUNT(*)                         AS uses,
  COUNT(DISTINCT b.user_id)        AS donors,
  ROUND(AVG(p.n), 1)               AS avg_donations_of_users
FROM donorsearch.user_anon_bonus b
LEFT JOIN per_user p ON p.user_id = b.user_id
GROUP BY 1, 2
HAVING COUNT(*) >= 30
ORDER BY uses DESC
LIMIT 20;
```

Основные выводы:

- Монотонный рост выглядит убедительно, но он ожидаем при любой причинной схеме. Бонусы начисляются за донации, поэтому тот, кто сдает больше, копит и тратит больше бонусов. Это обратная причинность, а не эффект бонусов.

Выводы по дополнительным запросам:
- После первого бонуса донаций становится меньше, а не больше. У 66% доноров после бонуса донаций меньше, чем до него. Но это не значит, что бонус отпугивает. Скорее всего, бонус берут сразу после донации, и затем наступает естественный перерыв: между донациями цельной крови проходит не менее двух месяцев. Поэтому сам по себе спад он ничего не доказывает, но и роста вовлеченности не показывает.
- Контрольная группа: у доноров с бонусами возврат не выше, а часто ниже. Во всех трех корзинах доноры, уже получавшие бонусы, возвращаются реже. Возможно, бонусы чаще брали доноры с прошлым пиком активности, которые уже сворачивали донорство. Возможно и то, что бонус дает награду и снимает мотивацию продолжать.
- Самым популярным бонусом оказалась пицца, но брали ее доноры с наименьшим стажем (9,1). Дорогие по восприятию и нишевые бонусы (МИФ, Flowwow, Литрес, МТС Premium) выбирают более опытные доноры.

Ограничения:
- Обратная причинность. Бонусы зависят от числа донаций, поэтому корреляция не доказывает влияния.
- Нет даты получения бонуса. Мы видим только факт использования, а не момент, когда бонус стал доступен.
- Бонусы берут доноры, которые сами решили ими воспользоваться, и эти доноры исходно более вовлечены. Поэтому разница в активности между группами отражает в первую очередь различие между людьми, а не эффект бонусов.

Рекомендации:
- Не оценивать программу по тому, что использующие бонусы активнее: это в основном эффект отбора.
- Проверить А/В-тестом, а не по историческим данным: часть доноров получает напоминание о бонусе после донации, часть не получает.
- Привязать бонусы к следующей донации (за вторую и третью сдачу в течение полугода), а не только к накопленному числу. Сейчас, судя по данным, бонус вознаграждает прошлое поведение и не стимулирует возврат.
- Сфокусироваться на донорах с одной донацией: у них возврат ниже всего. Для этой группы стоит проверить другие механики, например напоминание о сроке следующей сдачи.
- Продвигать бонусы, интересные опытным донорам (книги, подписки, доставка), в них более высокая лояльность. Массовые бонусы (пицца) приносят много использований, но не удерживают.

```python
%%sql
-- Проверка полей

SELECT
  COUNT(*)                                        AS users,
  COUNT(*) FILTER (WHERE autho_vk IS NULL)        AS vk_null,
  COUNT(*) FILTER (WHERE autho_ok IS NULL)        AS ok_null,
  COUNT(*) FILTER (WHERE autho_tg IS NULL)        AS tg_null,
  COUNT(*) FILTER (WHERE autho_yandex IS NULL)    AS yandex_null,
  COUNT(*) FILTER (WHERE autho_google IS NULL)    AS google_null,
  COUNT(*) FILTER (WHERE email_is_specified IS TRUE) AS with_email,
  COUNT(*) FILTER (WHERE phone_specified IS TRUE)    AS with_phone
FROM donorsearch.user_anon_data;
```

```python
%%sql
-- Доноры и донации по каналам

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
per_user AS (SELECT user_id, COUNT(*) AS n FROM clean GROUP BY 1),
u AS (
  SELECT d.id,
         COALESCE(d.autho_vk, FALSE)     AS vk,
         COALESCE(d.autho_ok, FALSE)     AS ok,
         COALESCE(d.autho_tg, FALSE)     AS tg,
         COALESCE(d.autho_yandex, FALSE) AS yandex,
         COALESCE(d.autho_google, FALSE) AS google,
         COALESCE(p.n, 0)                AS n
  FROM donorsearch.user_anon_data d
  LEFT JOIN per_user p ON p.user_id = d.id
),
long AS (
  SELECT 'VK' AS channel, id, n FROM u WHERE vk
  UNION ALL SELECT 'Одноклассники', id, n FROM u WHERE ok
  UNION ALL SELECT 'Telegram',      id, n FROM u WHERE tg
  UNION ALL SELECT 'Яндекс',        id, n FROM u WHERE yandex
  UNION ALL SELECT 'Google',        id, n FROM u WHERE google
  UNION ALL SELECT 'Без соцсетей',  id, n FROM u
            WHERE NOT (vk OR ok OR tg OR yandex OR google)
)
SELECT
  channel,
  COUNT(*)                                                  AS users,
  ROUND(100.0 * COUNT(*) / (SELECT COUNT(*) FROM u), 1)     AS pct_of_all_users,
  COUNT(*) FILTER (WHERE n > 0)                             AS donors_with_donations,
  ROUND(100.0 * COUNT(*) FILTER (WHERE n > 0) / COUNT(*), 1) AS pct_with_donations,
  ROUND(AVG(n), 2)                                          AS avg_donations_all,
  ROUND(AVG(n) FILTER (WHERE n > 0), 2)                     AS avg_donations_active
FROM long
GROUP BY channel
ORDER BY users DESC;
```

```python
%%sql
-- Пересечения: сколько каналов на донора

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
per_user AS (SELECT user_id, COUNT(*) AS n FROM clean GROUP BY 1)
SELECT
  (COALESCE(d.autho_vk, FALSE)::int + COALESCE(d.autho_ok, FALSE)::int
 + COALESCE(d.autho_tg, FALSE)::int + COALESCE(d.autho_yandex, FALSE)::int
 + COALESCE(d.autho_google, FALSE)::int)            AS n_channels,
  COUNT(*)                                          AS users,
  ROUND(100.0 * COUNT(*) FILTER (WHERE p.n > 0) / COUNT(*), 1) AS pct_with_donations,
  ROUND(AVG(COALESCE(p.n, 0)), 2)                   AS avg_donations_all
FROM donorsearch.user_anon_data d
LEFT JOIN per_user p ON p.user_id = d.id
GROUP BY 1
ORDER BY 1;
```

```python
%%sql
-- Ровно один канал: убираем пересечения, чтобы один донор не попадал в несколько строк

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
per_user AS (SELECT user_id, COUNT(*) AS n FROM clean GROUP BY 1),
u AS (
  SELECT d.id, COALESCE(p.n, 0) AS n,
         COALESCE(d.autho_vk, FALSE)::int AS vk, COALESCE(d.autho_ok, FALSE)::int AS ok,
         COALESCE(d.autho_tg, FALSE)::int AS tg, COALESCE(d.autho_yandex, FALSE)::int AS yandex,
         COALESCE(d.autho_google, FALSE)::int AS google
  FROM donorsearch.user_anon_data d
  LEFT JOIN per_user p ON p.user_id = d.id
)
SELECT
  CASE WHEN vk+ok+tg+yandex+google = 0 THEN 'Без соцсетей'
       WHEN vk+ok+tg+yandex+google > 1 THEN 'Несколько каналов'
       WHEN vk = 1 THEN 'Только VK'
       WHEN ok = 1 THEN 'Только Одноклассники'
       WHEN tg = 1 THEN 'Только Telegram'
       WHEN yandex = 1 THEN 'Только Яндекс'
       ELSE 'Только Google' END                         AS channel_group,
  COUNT(*)                                              AS users,
  ROUND(100.0 * COUNT(*) FILTER (WHERE n > 0) / COUNT(*), 1) AS pct_with_donations,
  ROUND(AVG(n), 2)                                      AS avg_donations_all,
  ROUND(AVG(n) FILTER (WHERE n > 0), 2)                 AS avg_donations_active
FROM u
GROUP BY 1
ORDER BY users DESC;
```

Основные выводы:
- По объему лидирует VK, но это основа массовой базы, а не качественный канал. VK и "без соцсетей" вместе дают 90,5% базы. Сумма долей больше 100%, потому что каналы пересекаются.
- Малые каналы дают более высокую конверсию и вовлеченность. Telegram (39,2% с донациями), Яндекс (34,0%) и Google (24,3%) превосходят VK (16,4%) и "без соцсетей" (11,8%) по доле тех, кто сдал кровь. В среднем на пользователя разрыв еще больше: 4,20 у Telegram против 0,85 у VK. Telegram и Яндекс малы по объему (0,3% и 1,9%), поэтому для общей картины их вклад ограничен.

Выводы по дополнительным запросам:
- Главный сигнал в данных: число привязанных каналов. Виден резкий скачок: с 1 до 2 каналов доля доноров вырастает с 15% до 56%, а среднее значение почти в 9 раз. Но заметно, что 0 и 1 канал почти не различаются (0,62 против 0,70).
- Чистое сравнение (ровно один канал) показывает, что сам канал почти ничего не определяет. Когда убраны пересечения, преимущество Telegram, Яндекса и Google почти исчезает. Среди активных доноров все одноканальные группы лежат в узком диапазоне 4,3-5,4 донации. А высокие показатели малых каналов в первой таблице создавались в основном доноры с несколькими привязанными каналами, которые попали сразу в несколько строк. Например, у Telegram 890 пользователей, но только 471 из них с одним каналом: больше чем у половины привязаны и другие.

Ограничения:
- Это не источник привлечения. Мы видим способ входа, а не то, как человек узнал о проекте.
- Самоотбор. Человек, привязавший 2-3 канала, уже вовлечен в сервис. Связь «больше каналов - больше донаций» не доказывает, что привязка каналов повышает активность. Возможна и обратная связь: донор с историей донаций чаще заходит и привязывает аккаунты.
- Стаж регистрации не учтен. Каналы появлялись в разные годы, а более ранние пользователи имели больше времени на донации.
- Малые выборки. Telegram (471 в чистой группе), значения в 4 и 5 каналов (131 и 44) дают нестабильные оценки.

Рекомендации:
- Не считать Telegram, Яндекс или Google «качественными каналами» только по первой таблице: после очистки от пересечений разница мала. Решения об инвестициях в каналы по этим цифрам принимать не стоит.
- Рассматривать привязку второго и последующих каналов как индикатор вовлеченности. Можно использовать это в сегментации: пользователей с 2+ каналами считать высоковероятными донорами, а в коммуникациях с остальными добиваться первой донации.
- Главный резерв роста лежит в массовой базе VK и «без соцсетей» (90% пользователей, 12-16% конверсии). Даже небольшой рост конверсии в первую донацию в этой группе даст больше, чем рост в малых каналах.
- Проверить, увеличивает ли активность предложение привязать второй канал (A/B-тест), а не предполагать это по данным.

## 7. Активность однократных доноров и повторных доноров

```python
%%sql
-- Доли сегментов (по визитам и по донациям)

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
per_user AS (
  SELECT user_id,
         COUNT(*)                      AS donations,
         COUNT(DISTINCT donation_date) AS visits,
         MIN(donation_date)            AS first_don,
         MAX(donation_date)            AS last_don
  FROM clean
  GROUP BY 1
)
SELECT
  COUNT(*)                                                    AS donors,
  COUNT(*) FILTER (WHERE donations = 1)                       AS single_by_donations,
  COUNT(*) FILTER (WHERE visits = 1)                          AS single_by_visits,
  COUNT(*) FILTER (WHERE visits >= 2)                         AS repeat_by_visits,
  ROUND(100.0 * COUNT(*) FILTER (WHERE visits = 1) / COUNT(*), 1) AS pct_single_visits
FROM per_user;
```

```python
%%sql
-- Интервалы между визитами у повторных доноров

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
visits AS (
  SELECT DISTINCT user_id, donation_date FROM clean
),
gaps AS (
  SELECT user_id,
         donation_date - LAG(donation_date) OVER (PARTITION BY user_id ORDER BY donation_date) AS gap_days
  FROM visits
)
SELECT
  COUNT(*)                                                    AS intervals,
  PERCENTILE_CONT(0.25) WITHIN GROUP (ORDER BY gap_days)      AS p25,
  PERCENTILE_CONT(0.5)  WITHIN GROUP (ORDER BY gap_days)      AS median_gap,
  PERCENTILE_CONT(0.75) WITHIN GROUP (ORDER BY gap_days)      AS p75,
  ROUND(100.0 * COUNT(*) FILTER (WHERE gap_days <= 120) / COUNT(*), 1) AS pct_within_120d,
  ROUND(100.0 * COUNT(*) FILTER (WHERE gap_days > 365)  / COUNT(*), 1) AS pct_over_year
FROM gaps
WHERE gap_days IS NOT NULL;
```

```python
%%sql
-- Профиль сегментов: пол, возраст, бонусы, каналы, профиль

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
per_user AS (
  SELECT user_id,
         COUNT(DISTINCT donation_date) AS visits,
         MIN(donation_date)            AS first_don
  FROM clean
  GROUP BY 1
),
seg AS (
  SELECT p.user_id,
         CASE WHEN p.visits = 1 THEN 'однократные' ELSE 'повторные' END AS segment,
         u.gender, u.birth_date, u.blood_type, u.region,
         u.count_bonuses_taken,
         u.email_is_specified, u.phone_specified,
         (COALESCE(u.autho_vk, FALSE)::int + COALESCE(u.autho_ok, FALSE)::int
        + COALESCE(u.autho_tg, FALSE)::int + COALESCE(u.autho_yandex, FALSE)::int
        + COALESCE(u.autho_google, FALSE)::int) AS n_channels
  FROM per_user p
  JOIN donorsearch.user_anon_data u ON u.id = p.user_id
  WHERE p.first_don <= DATE '2022-11-28'
)
SELECT
  segment,
  COUNT(*)                                                              AS donors,
  ROUND(100.0 * COUNT(*) FILTER (WHERE gender IS NOT NULL) / COUNT(*), 1) AS pct_gender_known,
  ROUND(100.0 * COUNT(*) FILTER (WHERE gender ILIKE 'Мужской')
        / NULLIF(COUNT(*) FILTER (WHERE gender IS NOT NULL), 0), 1)     AS pct_male_of_known,
  ROUND(AVG(EXTRACT(YEAR FROM AGE(DATE '2023-11-28', birth_date)))
        FILTER (WHERE birth_date BETWEEN '1930-01-01' AND '2010-01-01'), 1) AS avg_age,
  ROUND(100.0 * COUNT(*) FILTER (WHERE birth_date BETWEEN '1930-01-01' AND '2010-01-01') / COUNT(*), 1) AS pct_age_known,
  ROUND(100.0 * COUNT(*) FILTER (WHERE blood_type IS NOT NULL) / COUNT(*), 1) AS pct_blood_known,
  ROUND(100.0 * COUNT(*) FILTER (WHERE count_bonuses_taken > 0) / COUNT(*), 1) AS pct_took_bonus,
  ROUND(100.0 * COUNT(*) FILTER (WHERE n_channels >= 2) / COUNT(*), 1)   AS pct_2plus_channels,
  ROUND(100.0 * COUNT(*) FILTER (WHERE email_is_specified) / COUNT(*), 1) AS pct_email,
  ROUND(100.0 * COUNT(*) FILTER (WHERE phone_specified) / COUNT(*), 1)    AS pct_phone
FROM seg
GROUP BY 1;  
```

```python
%%sql
-- Тип и компонент первой донации

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
first_d AS (
  SELECT DISTINCT ON (user_id) user_id, donation_type, blood_class, donation_date
  FROM clean
  ORDER BY user_id, donation_date, id
),
v AS (SELECT user_id, COUNT(DISTINCT donation_date) AS visits FROM clean GROUP BY 1)
SELECT
  f.blood_class,
  f.donation_type,
  COUNT(*)                                                      AS donors,
  ROUND(100.0 * COUNT(*) FILTER (WHERE v.visits >= 2) / COUNT(*), 1) AS pct_repeat
FROM first_d f
JOIN v ON v.user_id = f.user_id
WHERE f.donation_date <= DATE '2022-11-28'
GROUP BY 1, 2
HAVING COUNT(*) >= 50
ORDER BY donors DESC;
```

```python
%%sql
-- Доля повторных по году первой донации

WITH clean AS (
  SELECT DISTINCT ON (user_id, donation_date, donation_type, blood_class) *
  FROM donorsearch.donation_anon
  WHERE donation_status = 'Принята'
    AND donation_date >= '1990-01-01'
    AND donation_date <= '2023-11-28'
  ORDER BY user_id, donation_date, donation_type, blood_class,
           confirmation DESC NULLS LAST, id
),
per_user AS (
  SELECT user_id, MIN(donation_date) AS first_don,
         COUNT(DISTINCT donation_date) AS visits
  FROM clean GROUP BY 1
)
SELECT EXTRACT(YEAR FROM first_don)::int AS first_year,
       COUNT(*)                          AS donors,
       ROUND(100.0 * COUNT(*) FILTER (WHERE visits >= 2) / COUNT(*), 1) AS pct_repeat
FROM per_user
WHERE first_don >= '2015-01-01' AND first_don <= '2022-11-28'
GROUP BY 1
ORDER BY 1;
```

Основные выводы:

- Побуждать к заполнению профиля сразу после первой донации: это совпадает с признаком вовлеченности, но причинный эффект нужно проверить.

## Планируемые и фактические донации

```python
%%sql
-- Изучим таблицу

SELECT * FROM donorsearch.donation_plan ORDER BY user_id, plan_date LIMIT 5;
```

```python
%%sql
-- Статусы планов по прошедшим датам

WITH d AS (
  SELECT user_id, plan_date, donation_type, blood_class
  FROM donorsearch.donation_plan
  GROUP BY 1, 2, 3, 4
  HAVING COUNT(*) > 1
)
SELECT
  COUNT(*)                                       AS dup_groups,
  COUNT(*) FILTER (WHERE n_conf > 1)             AS differ_confirmation,
  COUNT(*) FILTER (WHERE n_date > 1)             AS differ_donation_date,
  COUNT(*) FILTER (WHERE n_place > 1)            AS differ_place
FROM (
  SELECT p.user_id, p.plan_date, p.donation_type, p.blood_class,
         COUNT(DISTINCT confirmation)   AS n_conf,
         COUNT(DISTINCT donation_date)  AS n_date,
         COUNT(DISTINCT donation_place) AS n_place
  FROM donorsearch.donation_plan p
  JOIN d USING (user_id, plan_date, donation_type, blood_class)
  GROUP BY 1, 2, 3, 4
) t;
```

```python
%%sql
-- Реальная донация вокруг запланированной даты

WITH plan_users AS (
  SELECT DISTINCT user_id FROM donorsearch.donation_plan
),
clean AS (
  SELECT DISTINCT ON (a.user_id, a.donation_date, a.donation_type, a.blood_class)
         a.user_id, a.donation_date
  FROM donorsearch.donation_anon a
  JOIN plan_users pu ON pu.user_id = a.user_id
  WHERE a.donation_status = 'Принята'
    AND a.donation_date >= '2020-10-01'
    AND a.donation_date <= '2023-11-28'
  ORDER BY a.user_id, a.donation_date, a.donation_type, a.blood_class,
           a.confirmation DESC NULLS LAST, a.id
),
visits AS (
  SELECT DISTINCT user_id, donation_date FROM clean
),
p AS (
  SELECT id, user_id, donation_date AS planned_date, confirmation
  FROM donorsearch.donation_plan
  WHERE donation_date >= plan_date
    AND donation_date <= DATE '2023-11-21'
),
f AS (
  SELECT p.id, p.confirmation,
         COUNT(v.donation_date) > 0 AS fulfilled
  FROM p
  LEFT JOIN visits v
         ON v.user_id = p.user_id
        AND v.donation_date BETWEEN p.planned_date - 7 AND p.planned_date + 7
  GROUP BY p.id, p.confirmation
)
SELECT
  confirmation,
  COUNT(*)                                                       AS plans,
  COUNT(*) FILTER (WHERE fulfilled)                              AS with_real_donation,
  ROUND(100.0 * COUNT(*) FILTER (WHERE fulfilled) / COUNT(*), 1) AS pct_with_real
FROM f
GROUP BY 1
ORDER BY plans DESC;
```

```python
%%sql
-- Общая эффективность планирования

WITH plan_users AS (
  SELECT DISTINCT user_id FROM donorsearch.donation_plan
),
clean AS (
  SELECT DISTINCT ON (a.user_id, a.donation_date, a.donation_type, a.blood_class)
         a.user_id, a.donation_date
  FROM donorsearch.donation_anon a
  JOIN plan_users pu ON pu.user_id = a.user_id
  WHERE a.donation_status = 'Принята'
    AND a.donation_date >= '2020-10-01'
    AND a.donation_date <= '2023-11-28'
  ORDER BY a.user_id, a.donation_date, a.donation_type, a.blood_class,
           a.confirmation DESC NULLS LAST, a.id
),
visits AS (
  SELECT DISTINCT user_id, donation_date FROM clean
),
p AS (
  SELECT DISTINCT ON (user_id, plan_date, donation_type, blood_class)
         user_id, plan_date, donation_date AS planned_date, confirmation
  FROM donorsearch.donation_plan
  WHERE donation_date >= plan_date
    AND donation_date <= DATE '2023-11-21'
  ORDER BY user_id, plan_date, donation_type, blood_class, id DESC
),
f AS (
  SELECT p.user_id, p.plan_date, p.planned_date, p.confirmation,
         COUNT(v.donation_date) > 0 AS fulfilled
  FROM p
  LEFT JOIN visits v
         ON v.user_id = p.user_id
        AND v.donation_date BETWEEN p.planned_date - 7 AND p.planned_date + 7
  GROUP BY p.user_id, p.plan_date, p.planned_date, p.confirmation
)
SELECT
  COUNT(*)                                                       AS plans,
  COUNT(*) FILTER (WHERE fulfilled)                              AS fulfilled,
  ROUND(100.0 * COUNT(*) FILTER (WHERE fulfilled) / COUNT(*), 1) AS pct_fulfilled,
  COUNT(*) FILTER (WHERE confirmation <> 'планирование отменено') AS not_cancelled,
  ROUND(100.0 * COUNT(*) FILTER (WHERE fulfilled AND confirmation <> 'планирование отменено')
        / NULLIF(COUNT(*) FILTER (WHERE confirmation <> 'планирование отменено'), 0), 1) AS pct_fulfilled_of_active
FROM f;
```

```python
%%sql
-- Типичное опережение: за сколько дней создают план

SELECT
  CASE WHEN donation_date - plan_date <= 3   THEN '0-3 дня'
       WHEN donation_date - plan_date <= 14  THEN '4-14 дней'
       WHEN donation_date - plan_date <= 60  THEN '15-60 дней'
       WHEN donation_date - plan_date <= 120 THEN '61-120 дней'
       ELSE 'более 120 дней' END AS lead_time,
  COUNT(*) AS plans,
  ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 1) AS pct
FROM donorsearch.donation_plan
WHERE donation_date >= plan_date
GROUP BY 1
ORDER BY MIN(donation_date - plan_date);
```

- Статус плана хорошо отражает факт, но с оговоркой. "Запланированная донация добавлена" это статус выполнения. У этой группы 66,3% реальных совпадений, у остальных 1-14%.
